```python
import json
```

```python
"""Take first element of similarity dataset"""
with open("similarity_data.json") as data_file:
    data = json.load(data_file)

data[0]
```

```python
"""Based on a threshold, decide if it is synthetic or not"""
def predict_synth_code(data, threshold = 0.7):
    false_positives = 0
    false_negatives = 0
    correct_predictions = 0
    for elem in data:
        is_synth_prediction = elem['similarity'] > threshold
        if is_synth_prediction and elem['is_human']:
            false_positives += 1
        elif not is_synth_prediction and not elem['is_human']:
            false_negatives += 1
        else:
            correct_predictions += 1

    print(f'Correct {(correct_predictions/len(data))*100}%')
    print(f'False positives {(false_positives/len(data))*100}%')
    print(f'False negatives {(false_negatives/len(data))*100}%')
```

```python
predict_synth_code(data, 0.9999)
```
