```python
import json
from difflib import SequenceMatcher
import numpy as np
```

```python
"""Take first element of rewritten dataset"""
with open("rewritten_data.json") as data_file:
    data = json.load(data_file)

data[0]
```

```python
def normalized_edit_distance(s1, s2):
    """Compute normalized edit distance (0 = identical, 1 = completely different)"""
    ratio = SequenceMatcher(None, s1, s2).ratio()
    return 1 - ratio

def compute_similarity(original, rewrites):
    """Compute average similarity between original and rewrites"""
    
    if not rewrites:
        return 0.0
        
     # Lower distance = higher similarity for synthetic code
    distances = [normalized_edit_distance(original, r) for r in rewrites]
    # Convert to similarity score (invert)
    return 1 - np.mean(distances)
```

```python
compute_similarity(data[0]['code'], data[0]['rewrites'])
```

```python
i = 0
data_with_similarities = []
for elem in data:
    elem['similarity'] = compute_similarity(elem['code'], elem['rewrites'])
    data_with_similarities.append(elem)
    i += 1
    print(f"Generated {i} of {len(data)}")
```

```python
with open("similarity_data.json", "w") as data_file:
    json.dump(data_with_similarities, data_file, indent=2)
```
